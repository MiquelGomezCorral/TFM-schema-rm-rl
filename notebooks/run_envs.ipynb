```python
from maikol_utils.file_utils import list_dir_files
from pathlib import Path
import dotenv

from scripts import generate_rm
from src.config import Configuration


dotenv.load_dotenv()

files = list_dir_files('../examples/', recursive=True)
files = [Path(f) for f in files if f.endswith('.json')]
files
```

```python
from maikol_utils.file_utils import load_json

experiments = [load_json(f) for f in files]
experiments
```

```python
from maikol_utils.print_utils import print_separator
for exp in experiments:
    print_separator(f"Experiment: {exp['env']}", sep_type='START')
    CONFIG = Configuration(
        environment=Path('../') / exp['env_description'],
        tasks=exp['tasks'],
        output=f"{exp['env']}.md",
        rm_critic = False,
        task_critic = False
    )


    generate_rm(
        CONFIG
    )
    print_separator(f"Experiment: {exp}", sep_type='END')
```
